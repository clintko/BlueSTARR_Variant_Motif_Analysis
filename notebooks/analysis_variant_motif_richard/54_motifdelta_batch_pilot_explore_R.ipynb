```r
fun_parse_motif = function(dat, col = "Motif_Name") {
    dat = dat %>% dplyr::mutate(
        ### deduplicate exact "X X" case
        "{col}" := str_replace(.data[[col]], "^(\\S+)\\s+\\1$", "\\1"),
    
        ### Motif_Index: take everything before the first ':'
        Motif_Index = str_extract(.data[[col]], "^[^:]+"),
    
        ### Motif_Label: take everything after the first ':'
        Motif_Label = str_replace(.data[[col]], "^[^:]+:", "")
    )
    return(dat)
}
```

**Export figures**

```r
fun_save_plot = function(plt_export, txt_prefix, num_height, num_width){
    formats = c("png", "pdf", "svg")

    for (fmt in formats){
        txt_fpath = paste0(txt_prefix, ".", fmt)

        if (fmt == "png") {
            ggsave(
                filename = txt_fpath,
                plot     = plt_export,
                width    = num_width,
                height   = num_height,
                dpi      = 300
            )

        } else if (fmt == "pdf") {
            ggsave(
                filename = txt_fpath,
                plot     = plt_export,
                width    = num_width,
                height   = num_height,
                device   = grDevices::cairo_pdf
            )

        } else if (fmt == "svg") {
            ggsave(
                filename = txt_fpath,
                plot     = plt_export,
                width    = num_width,
                height   = num_height
            )
        }
    }
}
```

```r
plt_export = gpt_hist_motif_delta_top
num_height = 5
num_width  = 10
txt_prefix = "./fig.delta_motif.histogram.pilot_top"
fun_save_plot(plt_export, txt_prefix, num_height, num_width)
```

## Explore: heatmap of motif gain/loss events across variants

```r
dat_event_full      = dat_motifdelta_top_event_arrange
dat_summary_motif   = dat_motifdelta_top_motif_arrange
dat_summary_variant = dat_motifdelta_top_variant_import
```

```r
### Choose which variants/motifs to plot
n_var   = 200
n_motif = 100

### pick variants: most events
vec_txt_variant = dat_summary_variant %>%
    dplyr::mutate(Total = Count_Gain + Count_Loss) %>%
    dplyr::arrange(desc(Total)) %>%
    dplyr::slice_head(n = n_var) %>%
    dplyr::pull(Variant_ID)

### pick motifs: most gain/loss (choose one criterion or Total)
vec_txt_motif = dat_summary_motif %>%
    dplyr::mutate(Total = Count_Gain + Count_Loss) %>%
    dplyr::arrange(desc(Total)) %>%
    dplyr::slice_head(n = n_motif) %>%
    dplyr::pull(Motif_Name)

### filter events to subset
dat_event_sub = dat_event_full %>%
    dplyr::filter(
        Variant_ID %in% vec_txt_variant, 
        Motif_Name %in% vec_txt_motif
    )

cut_prob = 0.7
cut_gain = quantile(dat_event_sub$Delta[dat_event_sub$Event_Type == "Gain"], probs = cut_prob, na.rm = TRUE)
cut_loss = quantile(dat_event_sub$Delta[dat_event_sub$Event_Type == "Loss"], probs = 1 - cut_prob, na.rm = TRUE)

### Build gain and loss matrices (wide format)
mat_event_gain = dat_event_sub %>%
    dplyr::filter(Event_Type == "Gain", Delta >= cut_gain) %>%
    dplyr::select(Variant_ID, Motif_Name, Delta) %>%
    tidyr::pivot_wider(names_from = Motif_Name, values_from = Delta, values_fill = NA_real_) %>%
    column_to_rownames("Variant_ID") %>%
    as.matrix()

mat_event_loss <- dat_event_sub %>%
    dplyr::filter(Event_Type == "Loss", Delta <= cut_loss) %>%
    #dplyr::mutate(DeltaAbs = abs(Delta)) %>%
    #dplyr::select(Variant_ID, Motif_Name, DeltaAbs) %>%
    #tidyr::pivot_wider(names_from = Motif_Name, values_from = DeltaAbs, values_fill = NA_real_) %>%
    dplyr::select(Variant_ID, Motif_Name, Delta) %>%
    tidyr::pivot_wider(names_from = Motif_Name, values_from = Delta, values_fill = NA_real_) %>%
    column_to_rownames("Variant_ID") %>%
    as.matrix()
```

```r
drop_sparse <- function(mat, min_non_na = 2) {
  mat <- mat[rowSums(!is.na(mat)) >= min_non_na, , drop = FALSE]
  mat <- mat[, colSums(!is.na(mat)) >= min_non_na, drop = FALSE]
  return(mat)
}

mat_event_gain2 = drop_sparse(mat_event_gain, min_non_na = 2)
mat_event_loss2 = drop_sparse(mat_event_loss, min_non_na = 2)

mat_gain_plot = mat_event_gain2
mat_gain_plot[is.na(mat_gain_plot)] = 0

mat_loss_plot = mat_event_loss2
mat_loss_plot[is.na(mat_loss_plot)] = 0
```

```r
vec_txt_color = colorRampPalette(brewer.pal(9, "Reds"))(100)

plt = pheatmap(
    mat_gain_plot,
    color = vec_txt_color,
    border_color = NA,
    clustering_distance_rows = "correlation",
    clustering_distance_cols = "correlation",
    cluster_rows = TRUE,
    cluster_cols = TRUE,
    show_rownames = FALSE,
    #fontsize_row = 6,
    fontsize_col = 8,
    main = sprintf("Strong Gain Δ (>= %.3f; q=%.2f)", cut_gain, cut_prob),
    silent = TRUE
)

plt_labeled = gridExtra::arrangeGrob(
    plt$gtable,
    left = grid::textGrob("Variants", rot = 90, gp = grid::gpar(fontsize = 15)),
    bottom = grid::textGrob("Motifs", gp = grid::gpar(fontsize = 15))
)

plt_heatmap_gain_subset = plt_labeled
options(repr.plot.height=10, repr.plot.width=10)
grid::grid.newpage()
grid::grid.draw(plt_labeled)
```

```r
vec_txt_color = colorRampPalette(brewer.pal(9, "Blues"))(100)
vec_txt_color = rev(vec_txt_color)

plt = pheatmap(
    mat_loss_plot,
    color = vec_txt_color,
    border_color = NA,
    clustering_distance_rows = "correlation",
    clustering_distance_cols = "correlation",
    cluster_rows = TRUE,
    cluster_cols = TRUE,
    show_rownames = FALSE,
    #fontsize_row = 6,
    fontsize_col = 8,
    main = sprintf("Strong Loss Δ (<= %.3f; q=%.2f)", cut_loss, cut_prob),
    silent = TRUE
)

plt_labeled = gridExtra::arrangeGrob(
    plt$gtable,
    left = grid::textGrob("Variants", rot = 90, gp = grid::gpar(fontsize = 15)),
    bottom = grid::textGrob("Motifs", gp = grid::gpar(fontsize = 15))
)

plt_heatmap_loss_subset = plt_labeled
options(repr.plot.height=10, repr.plot.width=10)
grid::grid.newpage()
grid::grid.draw(plt_labeled)
```

**Export figures**

```r
fun_save_grob <- function(plt, filename, width = 10, height = 10) {
    grDevices::cairo_pdf(filename, width = width, height = height)
    grid::grid.newpage()
    grid::grid.draw(plt)
    grDevices::dev.off()
}
```
